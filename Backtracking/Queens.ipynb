## Rueens problem
You have a n*n grid on which you have to place n Rueens, where none of the queens can atack each other, you have to return all possible disposition of queens where this condition is meet.

Use backtracking for the implementation of the solution, you have tu return the 

In python a stack is just a simple list, and uses the following methods
s.append()
s.pop()

```python
def canPlaceQueen(current_grid, i, j, n):
    #Row
    for c in range(n):
        if current_grid[i][c]=='R':
            return False

    #Column
    for r in range(n):
        if current_grid[r][j]=='R':
            return False

    r = i
    c = j
    while r<n and c<n:
        if current_grid[r][c] == 'R':
            return False
        r +=1
        c +=1

    r = i
    c = j
    while r>=0 and c<n:
        if current_grid[r][c] == 'R':
            return False
        r -=1
        c +=1

    r = i
    c = j
    while r>=0 and c>=0:
        if current_grid[r][c] == 'R':
            return False
        r -=1
        c -=1

    r = i
    c = j
    while r<n and c>=0:
        if current_grid[r][c] == 'R':
            return False
        r +=1
        c -=1

    return True    
```

```python
class QueensPartialSolution:
 def __init__(self, current_grid, current_row):
    self.current_grid = current_grid
    self.current_row = current_row 
```

```python
def queens(n):
    solutions =[]
    current_grid = ["-" * n for _ in range(n)]
    root_sol = QueensPartialSolution(current_grid,0)
    s =[] #
    s.append(root_sol)

    while len(s) != 0:
        partialSolution = s.pop()

        if partialSolution.current_row == n:
            solutions.append(partialSolution.current_grid)
            continue

        for c in range(n):
            if(not canPlaceQueen(partialSolution.current_grid, partialSolution.current_row, c, n)):
                continue

            childSolution = copy.deepcopy(partialSolution)
            # childSolution.current_grid[partialSolution.current_row][c]='Q' This is not possible
            row = childSolution.current_grid[partialSolution.current_row]   
            list_row = list(row)
            list_row[c]='R'
            new_row = "".join(list_row)
            childSolution.current_grid[partialSolution.current_row] = new_row 
            childSolution.current_row +=1
            s.append(childSolution)
            
    return solutions
    print(len(solutions))

a=''''
    for i in range(len(solutions)):
        grid = solutions[i] 
        for row in grid:
            print(row)
        print("\n")
'''
```
