```python
import math
def d(A,B):
  return math.sqrt(math.pow(A[0]-B[0],2)+math.pow(A[1]-B[1],2))
```

```python
def closest2dBruteForce(OX):
    min_dist = float('inf')
    A,B = None, None
    n= len(OX)
    for i in range (n):
      for j in range (i+1, n):
        distance = d(OX[i],OX[j])
        if(distance < min_dist):
          min_dist = distance
          A,B = OX[i],OX[j]

    return min_dist, [A,B]
```

```python
def strip2d(strip):
  min_dist = float('inf')
  A,B = None,None
  m = len(strip)
  for i in range(m):
    for j in range(i+1,m):
      distance = d(strip[i], strip[j])
      if(distance < min_dist):
        min_dist = distance
        A,B = strip[i],strip[j]

  return min_dist, [A,B]
```

```python
def findClosest(OX, OY):
  n = len(OX)
  if n <= 3:
    return closest2dBruteForce(OX)

  mid = n//2
  midPoint = OX[mid]

  OXL = OX[:mid]
  OXR = OX[mid:]
  OYL = [p for p in OY if p[0] <= midPoint[0]] #Here is the x
  OYR = [p for p in OY if p[0] > midPoint[0]]

  minL, closestL = findClosest(OXL, OYL)
  minR, closestR = findClosest(OXR, OYR)

  if minL < minR:
    minA = minL
    closestA = closestL
  else:
    minA = minR
    closestA = closestR

  S = [p for p in OY if abs(p[0] - midPoint[0]) < minA]
  minS, closestS = strip2d(S)

  if minA < minS:
    return minA,closestA
  else:
    return minS, closestS
```

```python
class Point:
    x = None
    y = None
    z = None

    def __init__(self, coor_x, coor_y, coor_z):
        self.x = coor_x
        self.y = coor_y
        self.z = coor_z

    def plot(self, tag, ax=None):
        if ax is not None:
            ax.scatter(self.x, self.y, self.z)
            ax.text(self.x, self.y, self.z, tag)
        else:
            # Fallback for standard 2D plots if used
            plt.scatter(self.x, self.y)
            plt.text(self.x, self.y, tag)
```

```python
def d3d(A,B):
  return math.sqrt(
      math.pow(A[0]-B[0],2)+
      math.pow(A[1]-B[1],2)+
      math.pow(A[2]-B[2],2)
      )
```

```python
def closest3dBruteForce(OX):
    min_dist = float('inf')
    A,B = None, None
    n= len(OX)
    for i in range (n):
      for j in range (i+1, n):
        distance = d3d(OX[i],OX[j])
        if(distance < min_dist):
          min_dist = distance
          A,B = OX[i],OX[j]

    return min_dist, [A,B]
```

```python
def strip2d(strip):
  min_dist = float('inf')
  A,B = None,None
  m = len(strip)
  for i in range(m):
    for j in range(i+1,m):
      distance = d(strip[i], strip[j])
      if(distance < min_dist):
        min_dist = distance
        A,B = strip[i],strip[j]

  return min_dist, [A,B]
```

```python
def stripZ3d(stripZ, minDist):
  n = len(stripZ)
  min_dist = float('inf')
  A,B = None, None
  for i in range(n):
    j = i+1
    while j < n and abs(stripZ[i][2] - stripZ[j][2]) < minDist:
      distance = d3d(stripZ[i], stripZ[j])
      if(distance < min_dist):
        min_dist = distance
        A,B = stripZ[i], stripZ[j]
      j += 1

  return min_dist, [A,B]
```

```python
def strip3d(stripY, stripZ):
  m = len(stripY)
  if m <= 3:
    return closest3dBruteForce(stripY)

  mid = m//2
  midPoint = stripY[mid]

  syl = stripY[:mid]
  syh = stripY[mid:]
  szl = [p for p in stripZ if p[1] <= midPoint[1]] ## Against the y
  szh = [p for p in stripZ if p[1] > midPoint[1]]

  minLow, closestLow = strip3d(syl,szl)
  minHigh, closestHigh = strip3d(syh, szh)

  if minLow < minHigh:
    minA = minLow
    closestA = closestLow
  else:
    minA = minHigh
    closestA = closestHigh

  strip = [p for p in stripZ if abs(p[1] - midPoint[1]) < minA]
  minS, closestS = stripZ3d(strip, minA)

  if minA < minS:
    return minA, closestA
  else:
    return minS, closestS
```

```python
def findClosest3d(OX, OY,OZ):
  n = len(OX)
  if n <= 3:
    return closest3dBruteForce(OX)

  mid = n//2
  midPoint = OX[mid]

  OXL = OX[:mid]
  OXR = OX[mid:]
  OYL = [p for p in OY if p[0] <= midPoint[0]] #Here we're comparing against the x
  OYR = [p for p in OY if p[0] > midPoint[0]]
  OZL = [p for p in OZ if p[0] <= midPoint[0]]
  OZR = [p for p in OZ if p[0] > midPoint[0]]

  minL, closestL = findClosest3d(OXL, OYL,OZL)
  minR, closestR = findClosest3d(OXR, OYR, OZR)

  if minL < minR:
    minA = minL
    closestA = closestL
  else:
    minA = minR
    closestA = closestR

  stripY  = [p for p in OY if abs(p[0] - midPoint[0]) < minA]
  stripZ  = [p for p in OZ if abs(p[0] - midPoint[0]) < minA]

  minS, closestS = strip3d(stripY,stripZ)

  if minA < minS:
    return minA,closestA
  else:
    return minS, closestS
```
