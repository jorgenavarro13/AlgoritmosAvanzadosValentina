```python
import queue
import heapq
from collections import defaultdict
```

```python
def prim(connections,n,origin):

    distances = {}
    processed = {}
    predecesor = {}


    G = defaultdict(dict)
    for connection in connections:
        u,v,w = connection[0],connection[1],connection[2]
        G[u][v]=w
        G[v][u]=w
        processed[u]= False 
        processed[v]= False
        distances[u]= float('inf')
        distances[v]= float('inf')

    pq = []
    distances[origin]=0
    heapq.heapify(pq)
    heapq.heappush(pq, (0,origin))

    while (len(pq) > 0):
        u = heapq.heappop(pq)[1]

        if processed[u]:
            continue

        processed[u]=True

        for neighbor in G[u]:
            new_distance = G[u][neighbor]
            if new_distance < distances[neighbor] and not processed[neighbor]:
                distances[neighbor] = new_distance
                predecesor[neighbor] = u
                if(processed[neighbor]!=None):
                    heapq.heappush(pq,(distances[neighbor], neighbor))

    print(processed)
    print(predecesor)


```

```python
# [u, v, w] -:> List
edges = [
    ("A", "B", 3),
    ("A", "F", 5),
    ("A", "E", 6),
    ("B", "C", 1),
    ("B", "F", 4),
    ("C", "D", 6),
    ("C", "F", 4),
    ("D", "E", 8),
    ("D", "F", 5),
    ("E", "F", 2)
]

n = 6
```

```python
prim(edges,n,"A")
```
